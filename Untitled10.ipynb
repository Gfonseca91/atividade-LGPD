```python
import pandas as pd
#lib2
#lib3
```

```python
df_beneficio = pd.read_csv("/content/beneficios_saude.csv")
df_cliente = pd.read_csv("/content/clientes.csv")
df_compra = pd.read_csv("/content/compras.csv")
```

```python
display(df_beneficio.head(1))
display(df_cliente.head(1))
display(df_compra.head(1))
```

```python
# vamos juntar as tabelas em uma só
# temos que fazer em duas etapas
# sempre as tabelas dão merge em pares
df_beneficio_cliente = pd.merge(df_cliente,
                                  df_beneficio,
                                  on="cpf")
df_beneficio_cliente_compra = pd.merge(df_beneficio_cliente,
                                          df_compra,
                                          on="cliente_id")
display(df_beneficio_cliente_compra)
```

```python
# removendo dados sensíveis sem necessidade de exibição
df_dados_lgpd = df_beneficio_cliente_compra[["cliente_id",
                                             "idade",
                                             "cep",
                                             "plano_saude",
                                             "condicao_cronica",
                                             "transacao_id",
                                             "valor_gasto",
                                             "categoria"]]


#aplicação de máscara
df_dados_lgpd["regiao"] = df_dados_lgpd["cep"].str[:3] + "xx-xxx"
display(df_dados_lgpd)

#regra da faixa
bins = [0,25,35,75,120]
faixas = ["18-25", "26-35", "36-50", "60+"]
df_dados_lgpd["faixa etaria"] = pd.cut(df_dados_lgpd["idade"],
                                       bins=bins,
                                       labels=faixas)
display(df_dados_lgpd)
```
